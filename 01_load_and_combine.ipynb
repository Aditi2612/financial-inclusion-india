```python
all_data = []

for f in files:
    temp = pd.read_excel(f, header=5)
    temp = temp.dropna(axis=1, how='all')      # drop fully-empty columns
    all_data.append(temp)
    print(f"Loaded {f[-20:]} — {temp.shape}")

df = pd.concat(all_data, ignore_index=True)
print("COMBINED:", df.shape)
```

```python
df['Population Group - Utilization'] = df['Population Group - Utilization'].replace('SEMI URBAN', 'SEMI-URBAN')
print(df['Population Group - Utilization'].unique())
```

```python
df.groupby('Population Group - Utilization')['Credit Limit (in Rs Crores)'].sum()
```

```python
import matplotlib.pyplot as plt

accounts = df.groupby('Population Group - Utilization')['No. of Accounts'].sum()
accounts.plot(kind='bar')
plt.show()
```

```python
accounts = df.groupby('Population Group - Utilization')['No. of Accounts'].sum()
accounts.plot(kind='bar', color='steelblue')
plt.title('Total Bank Accounts by Population Group (2010-2026)')
plt.ylabel('Number of Accounts')
plt.xlabel('Population Group')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
credit = df.groupby('Population Group - Utilization')['Credit Limit (in Rs Crores)'].sum()
credit.plot(kind='bar', color='darkorange')
plt.title('Total Credit by Population Group (2010-2026)')
plt.ylabel('Credit Limit (Rs Crores)')
plt.xlabel('Population Group')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
credit_per_account = credit / accounts
print(credit_per_account)
```
